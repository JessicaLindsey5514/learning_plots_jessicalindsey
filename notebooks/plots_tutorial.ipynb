```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np




```

```python
spotify_df = pd.read_csv("../data/Cleaned_Spotify_2024_Global_Streaming_Data.csv")
spotify_df.head()
```

```python
#group by genre and calculate total average streams
avg_streams = spotify_df.groupby("Genre")["Total Streams (Millions)"].mean().sort_values(ascending=False)
avg_streams
```

```python
plt.bar(avg_streams.index, avg_streams.values)
plt.show()

```

```python
plt.bar(avg_streams.index, avg_streams.values)

plt.xlabel("Genre")
plt.ylabel("Average Total Streams (Millions)")
plt.title("Average Total Streams by Genre")

plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.bar(avg_streams.index, avg_streams.values)

plt.xlabel("Genre")
plt.ylabel("Average Total Streams (Millions)")
plt.title("Average Total Streams by Genre")
plt.xticks(rotation=45, ha="right")


plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.bar(avg_streams.index, avg_streams.values)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.xlabel("Genre")
plt.ylabel("Average Total Streams (Millions)")
plt.title("Average Total Streams by Genre")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
avg_artist_streams = spotify_df.groupby("Artist")["Total Streams (Millions)"].mean().sort_values(ascending=False)
avg_artist_streams
```

```python
plt.figure(figsize=[10,6])
colors = ["#a168c5"]*len(avg_artist_streams)


max_index = avg_artist_streams.idxmax()
highlight_position = avg_artist_streams.index.get_loc(max_index)
colors[highlight_position] = "#d512e7"

plt.bar(avg_artist_streams.index, avg_artist_streams.values, color = colors, width=0.8)
ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.xlabel("Artist")
plt.ylabel("Average Total Streams (Millions)")
plt.title("Average Total Streams by Artist")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()

plt.savefig('../plots/avgtotal_byartist.png')
plt.show()
```

```python
plt.figure(figsize=[10,6])

special_artist = "Ariana Grande"
colors = ["#f7146b" if artist == special_artist else "gray" for artist in avg_artist_streams.index]


plt.bar(avg_artist_streams.index, avg_artist_streams.values, color = colors, width=0.7)
ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.xlabel("Artist")
plt.ylabel("Average Total Streams (Millions)")
plt.title("Ariana Grande ranks second in Total Average Streams")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.scatter(
    spotify_df["Streams Last 30 Days (Millions)"],
    spotify_df["Total Streams (Millions)"],
    color = "#259e49",
    alpha= 0.8,
    s= 25
    )

plt.xlabel("Streams This Month")
plt.ylabel("Total Streams")
plt.title("No Apparent Correlation Between Monthly Listens and Total Streams")

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()

plt.savefig('../plots/monvstotalstm.png')
plt.show
```

```python
plt.figure(figsize=(10,6))

featured_genre = "Pop"
colors = ["#1B2F85" if genre == featured_genre else "#ff8196" for genre in spotify_df["Genre"]]


plt.scatter(
    spotify_df["Streams Last 30 Days (Millions)"],
    spotify_df["Total Streams (Millions)"],
    color = colors,
    alpha= 0.8,
    s= 25
    )

plt.xlabel("Streams This Month")
plt.ylabel("Total Streams")

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.set_title("Do Pop Songs Achieve Higher Monthly Listens or Total Streams?", fontsize= 18, pad=35)

for spine in ax.spines.values():
    spine.set_linewidth(.25)
    spine.set_alpha(.5)

plt.tight_layout()


plt.savefig('../plots/popMuzik.png')
plt.show


```

```python
yearly_avg = spotify_df.groupby("Release Year")["Total Streams (Millions)"].mean().round(2)
yearly_avg

```

```python
plt.figure(figsize = (10, 6))
plt.plot(
    yearly_avg.index, 
    yearly_avg.values,
    marker ="X",
    linestyle = "--",
    color = "#e76a57",
    linewidth = 2)

plt.xlabel("Release Year")
plt.ylabel("Average Total Streams")
plt.title("Did Listeners Stream Less in 2022?", fontsize=18, pad=20)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.axvline(
    2022,
    color = "#10c42e",
    linewidth = 1,
    alpha= .8)

ax.text(
    2022+0.05,
    2225,
    "Dip in Total Streams",
    color = "#420d88",
    fontsize= 12
)

plt.ylim(0, 3500)

plt.tight_layout()

plt.savefig('../plots/2022lesstreams.png')
plt.show()
```

```python
category = "Platform Type"
values= "Total Streams (Millions)"

platform_totals = spotify_df.groupby(category)[values].sum()
platform_totals
```

```python
plt.figure(figsize= (8,8))

plt.pie(
    platform_totals,
    labels=None,
    startangle= 100,
    colors=["#f5b6e7", "#5aeb7e"],
    autopct= "%1.1f%%"
    )

plt.title("Almost Even Split Between Premium and Free Users")

plt.legend(
    labels= platform_totals.index,
    loc="lower right",
    title = "Platform"
)

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8,8))

wedges, texts, autotext = ax.pie(
    platform_totals,
    startangle= 100,
    autopct= "%1.1f%%",
    colors=["#f5b6e7", "#5aeb7e"],
    )

ax.set_title(
    "Almost Even Split Between Premium and Free Users",
    fontsize= 28,
    color= "#204b74",
    pad= 10,
    loc="center"
    )

for autotext in autotext:
    autotext.set_fontsize(28)
    autotext.set_color("#5463ec")
    autotext.set_weight("bold")


ax.legend(
    wedges,
    platform_totals.index,
    loc="lower center",
    bbox_to_anchor = (0.5, 0.94),
    ncol= 2
)

plt.tight_layout()

plt.savefig('../plots/premvsfree.png')
plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.hist(
    spotify_df["Avg Stream Duration (Min)"],
    bins=25,
    color= "#5cca30",
    alpha= 0.75,
    edgecolor= "#8a6709"
    )

plt.xlabel("Average Stream Duration(Min)", color="#b92339", fontsize=17)
plt.ylabel("Number of Tracks", color= "#13725E", fontsize=17)
plt.title("Distribution of Average Stream Duration", fontsize=18, pad=20, color= "#990e1c")

ax= plt.gca()
for spine in ax.spines.values():
    spine.set_linewidth(0.5)
    spine.set_alpha(0.5)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)




plt.tight_layout()

plt.savefig('../plots/dist_avgstmdur.png')
plt.show()


```

```python
numeric_spotify_df = spotify_df.select_dtypes(include=[np.number])
corr = numeric_spotify_df.corr()

im = plt.imshow(corr, vmin=-1, vmax=1, cmap="coolwarm")
plt.colorbar(im, label= "Pearson Corellation")


plt.xticks(range(len(corr.columns)), corr.columns, rotation= 55, ha="right", color= "#360663")
plt.yticks(range(len(corr.columns)), corr.columns, color="#48097c")
plt.title("Numeric Data Heat Map", fontsize=20, pad=20, color= "#256905")

plt.savefig('../plots/numeric_heatmap.png')
plt.show()

```
